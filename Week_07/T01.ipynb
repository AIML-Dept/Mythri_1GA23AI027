## Exercise 1: Constant Function f(x) = 0

### Objective
To implement Deutsch's algorithm for the constant function f(x) = 0 and verify that the algorithm produces the output 0.

### Theory
A constant function gives the same output for every input. For f(x) = 0:

f(0) = 0 and f(1) = 0

In Deutsch's algorithm, the oracle does not apply any operation to the output qubit. After the final Hadamard gate, measurement of the control qubit gives 0, indicating that the function is constant.

### Observation
The measurement result is 0 for all 1024 shots. Therefore, the algorithm correctly identifies f(x) = 0 as a constant function.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from IPython.display import display

simulator = AerSimulator()

qc1 = QuantumCircuit(2, 1)

qc1.x(1)
qc1.h([0, 1])

# Oracle: f(x) = 0
# No operation required

qc1.h(0)
qc1.measure(0, 0)

display(qc1.draw("mpl"))

counts1 = simulator.run(qc1, shots=1024).result().get_counts()
print("Measurement Result:", counts1)
```
